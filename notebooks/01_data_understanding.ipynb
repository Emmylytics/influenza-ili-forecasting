#### 3. Load Dataset

```python
# Load dataset
df = load_ilinet()
```

#### 4. Dataset Overview

```python
# Check first 5 rows
df.head()
```

```python
# Check last 5 rows
df.tail()
```

```python
# Check dataset structure
print(f"Shape: {df.shape}, (Rows, Columns)")
print("\nData Types:")
df.info()
print("\nSummary Statistics:")
display(df.describe().round(2))
```

#### 5. Data Quality Assessment

```python
# Missing values summary
missing_summary = pd.DataFrame({
    "missing_values": df.isna().sum(),
    "percentage (%)": (df.isna().mean() * 100).round(2)
})
display(missing_summary)
```

```python
# Check for duplicate rows
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Count occurrences of "X" in each column
x_summary = (pd.DataFrame({
        "count_of_X": (df == "X").sum()
    })
    .query("count_of_X > 0")
    .sort_values("count_of_X", ascending=False)
)

display(x_summary)
```

#### 6. Temporal Coverage

```python
# Weekly records per year
year_counts = (
    df.groupby("YEAR")
      .size()
      .rename("weekly_records")
      .reset_index()
)

display(year_counts)
```

# Key Findings

The initial inspection of the ILINet dataset provides the following observations:

- The dataset contains **1,499 weekly influenza surveillance records** measured across **15 variables**.
- The data spans **1997 to 2026**, with **1997** and **2026** representing partial years. Most years contain **52 epidemiological weeks**, while **2025** contains **53 weeks**.
- No missing (`NaN`) values or duplicate records were identified.
- Several variables contain `"X"` values that require further investigation before preprocessing and modeling.
- Overall, the dataset appears well structured and is suitable for further cleaning and preparation.
